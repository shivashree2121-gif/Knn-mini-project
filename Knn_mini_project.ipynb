```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

# Load dataset
df = pd.read_csv("/content/KNNAlgorithmDataset.csv")

# Display dataset
print(df)
```

```python
# Features
X = df[["texture_mean", "perimeter_mean"]]

# Target / Label
y = df["diagnosis"]

# Create KNN model (k = 3)
model = KNeighborsClassifier(n_neighbors=3)

# Train model
model.fit(X, y)

# Get input from user
texture = float(input("Enter texture_mean (g): "))
mean = float(input("Enter perimeter_mean (g):"))

# Predict
dia = [[texture, mean]]
prediction = model.predict(dia)

print("Predicted Diagnosis:", prediction[0])
```

```python
from sklearn.metrics import accuracy_score
y_pred_train = model.predict(X)
accuracy = accuracy_score(y, y_pred_train)
print("Accuracy on training data:", accuracy*100)
```
